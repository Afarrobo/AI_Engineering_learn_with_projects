# Question 1: Dataset Loading and Initial Exploration

```python
# 1. Display all column names
print("Column Names:")
print(df.columns)

# 2. Dataset information
print("\nDataset Information:")
print(df.info())

# 3. Missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

# Question 3: Categorical Data Analysis

**1.Count the number of patients with heart disease (target = 1) and without heart disease (target = 0)**

```python
print("Heart Disease Count:")
print(df['target'].value_counts())
```

**2.Count the number of male (sex = 1) and female (sex = 0) patients**

```python
print("Gender Count:")
print(df['sex'].value_counts())
```

# Question 5: Data Filtering Using Conditions

**1.Extract all patients older than 50 years**

```python
older_than_50 = df[df['age'] > 50]

print(older_than_50)
```

**2.Extract patients who have heart disease and a cholesterol level greater than 240**

```python
heart_high_chol = df[(df['target'] == 1) & (df['chol'] > 240)]

print(heart_high_chol)
```

**3.Count how many patients satisfy both conditions**

```python
count = heart_high_chol.shape[0]

print("Number of patients:", count)
```
